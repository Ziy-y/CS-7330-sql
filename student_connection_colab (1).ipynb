```python
''' local anaconda connection code
import mysql.connector

# Connect to the existing MySQL database on your Mac
connection = mysql.connector.connect(
    host="127.0.0.1",
    port=3306,
    user="root",
    password="12345678",
    database="student"
)

cursor = connection.cursor()

# Query 1: Find all CS students
cursor.execute("""
SELECT ID, name
FROM student
WHERE dept_name = %s
""", ("CS",))

print("CS Students:")

for row in cursor.fetchall():
    print(row)


# Query 2: Find students advised by instructor 1101
cursor.execute("""
SELECT s_ID
FROM advisor
WHERE i_ID = %s
""", ("1101",))

print("\nStudents advised by instructor 1101:")

for row in cursor.fetchall():
    print(row)


cursor.close()
connection.close()
```
